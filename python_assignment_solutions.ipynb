## Section A – Basic Functions and Numbers

### 1. Even or Odd

```python
def check_even_odd(n):
    if n % 2 == 0:
        print(f"{n} is Even")
    else:
        print(f"{n} is Odd")

n = int(input("Enter a number: "))
check_even_odd(n)
```

### 2. Positive, Negative or Zero

```python
def check_number(n):
    if n > 0:
        print(f"{n} is Positive")
    elif n < 0:
        print(f"{n} is Negative")
    else:
        print(f"{n} is Zero")

n = float(input("Enter a number: "))
check_number(n)
```

### 3. Find Largest of Two Numbers

```python
def find_largest(a, b):
    if a > b:
        return a
    else:
        return b

a = float(input("Enter first number: "))
b = float(input("Enter second number: "))
print("Largest:", find_largest(a, b))
```

### 4. Find Largest of Three Numbers (no max())

```python
def find_largest(a, b, c):
    largest = a
    if b > largest:
        largest = b
    if c > largest:
        largest = c
    return largest

a = float(input("Enter first number: "))
b = float(input("Enter second number: "))
c = float(input("Enter third number: "))
print("Largest:", find_largest(a, b, c))
```

### 5. Sum of Natural Numbers (for loop)

```python
def sum_natural(n):
    total = 0
    for i in range(1, n + 1):
        total += i
    return total

n = int(input("Enter n: "))
print(f"Sum = {sum_natural(n)}")
```

### 6. Multiplication Table

```python
def multiplication_table(n):
    for i in range(1, 11):
        print(f"{n} x {i} = {n * i}")

n = int(input("Enter a number: "))
multiplication_table(n)
```

### 7. Factorial Using Function (loop, no recursion)

```python
def factorial(n):
    result = 1
    for i in range(1, n + 1):
        result *= i
    return result

n = int(input("Enter a number: "))
print("Factorial:", factorial(n))
```

### 8. Count Digits (while loop)

```python
def count_digits(n):
    n = abs(n)
    if n == 0:
        return 1
    count = 0
    while n > 0:
        n = n // 10
        count += 1
    return count

n = int(input("Enter a number: "))
print("Number of digits:", count_digits(n))
```

### 9. Reverse a Number

```python
def reverse_number(n):
    reversed_num = 0
    n = abs(n)
    while n > 0:
        digit = n % 10
        reversed_num = reversed_num * 10 + digit
        n = n // 10
    return reversed_num

n = int(input("Enter a number: "))
print("Reversed number:", reverse_number(n))
```

### 10. Prime Number

```python
def check_prime(n):
    if n < 2:
        print("Not Prime")
    else:
        for i in range(2,n):
            if n % i == 0:
                print("Not Prime")
                break
        else:
            print("Prime")
check_prime(int(input("Enter Number: ")))


```

## Section B – String-Based Assignments

### 11. Count Characters in a String (no len())

```python
def count_characters(text):
    count = 0
    for ch in text:
        if ch == " ":
            continue
        else:
            count += 1
    return count

text = input("Enter text: ")
print("Character count:", count_characters(text))
```

### 12. Count Vowels

```python
def count_vowels(text):
    vowels = "aeiouAEIOU"
    count = 0
    for ch in text:
        if ch in vowels:
            count += 1
    return count

text = input("Enter text: ")
print("Vowel count:", count_vowels(text))
```

### 13. Count Consonants

```python
def count_consonants(text):
    vowels = "aeiouAEIOU"
    count = 0
    for ch in text:
        if ch.isalpha() and ch not in vowels:
            count += 1
    return count

text = input("Enter text: ")
print("Consonant count:", count_consonants(text))
```

### 14. Count Vowels and Consonants

```python
def count_vowels_consonants(text):
    vowels = "aeiouAEIOU"
    v_count = 0
    c_count = 0
    for ch in text:
        if ch.isalpha():
            if ch in vowels:
                v_count += 1
            else:
                c_count += 1
    print(f"Vowels : {v_count}")
    print(f"Consonants : {c_count}")

text = input("Enter text: ")
count_vowels_consonants(text)
```

### 15. Reverse a String (loop and slicing)

```python
def reverse_string(text):
    # Approach 1: using a loop
    reversed_loop = ""
    for ch in text:
        reversed_loop = ch + reversed_loop

    # Approach 2: using slicing
    reversed_slice = text[::-1]

    return reversed_loop, reversed_slice

text = input("Enter text: ")
loop_result, slice_result = reverse_string(text)
print("Loop approach   :", loop_result)
print("Slicing approach:", slice_result)
```

### 16. Check Palindrome String

```python
def check_palindrome(text):
    if text == text[::-1]:
        print(f"{text} -> Palindrome")
    else:
        print(f"{text} -> Not Palindrome")

text = input("Enter text: ")
check_palindrome(text)
```

### 17. Count Words (no split())

```python
def count_words(text):
    if count_characters(text) == 0:
        return 0
    word_count = 1
    for ch in text:
        if ch == " ":
            word_count += 1
    return word_count

text = input("Enter a sentence: ")
print("Word count:", count_words(text))
```

### 18. Find Frequency of a Character (no count())

```python
def character_frequency(text, ch):
    freq = 0
    for c in text:
        if c == ch:
            freq += 1
    return freq

text = input("Enter text: ")
ch = input("Enter character to search: ")
print("Frequency:", character_frequency(text, ch))
```

### 19. Remove Spaces

```python
def remove_spaces(text):
    result = ""
    for ch in text:
        if ch != " ":
            result += ch
    return result

text = input("Enter text: ")
print("Without spaces:", remove_spaces(text))
```

### 20. Convert Lowercase to Uppercase

```python
def convert_uppercase(text):
    # First solve using upper()
    builtin_result = text.upper()

    # Loop-based solution using character processing
    manual_result = ""
    for ch in text:
        if "a" <= ch <= "z":
            manual_result += chr(ord(ch) - ord("a") + ord("A"))
        else:
            manual_result += ch

    return builtin_result, manual_result

text = input("Enter text: ")
builtin_res, manual_res = convert_uppercase(text)
print("Using upper()   :", builtin_res)
print("Using loop      :", manual_res)
```

## Section C – String + Loop Problems

### 21. Count Uppercase and Lowercase Characters

```python
def count_case(text):
    upper = lower = digits = spaces = 0
    for ch in text:
        if ch.isupper():
            upper += 1
        elif ch.islower():
            lower += 1
        elif ch.isdigit():
            digits += 1
        elif ch == " ":
            spaces += 1
    print(f"Uppercase : {upper}")
    print(f"Lowercase : {lower}")
    print(f"Digits : {digits}")
    print(f"Spaces : {spaces}")

text = input("Enter text: ")
count_case(text)
```

### 22. Find First Character (no text[0])

```python
def first_character(text):
    for ch in text:
        print(f"First character: {ch}")
        break

text = input("Enter text: ")
first_character(text)
```

### 23. Find Last Character

```python
def last_character(text):
    # Using slicing/indexing
    direct_result = text[-1]

    # Loop-based approach
    loop_result = None
    for ch in text:
        loop_result = ch

    print("Using text[-1] :", direct_result)
    print("Using loop     :", loop_result)

text = input("Enter text: ")
last_character(text)
```

### 26. Remove Vowels (no replace())

```python
def remove_vowels(text):
    vowels = "aeiouAEIOU"
    result = ""
    for ch in text:
        if ch not in vowels:
            result += ch
    return result

text = input("Enter text: ")
print("Without vowels:", remove_vowels(text))
```

### 27. Find Longest Word (no dict, no max())

```python
def find_longest_word(text):
    longest = ""
    current_word = ""
    for ch in text:
        if ch == " ":
            if count_characters(current_word) > count_characters(longest):
                longest = current_word
            current_word = ""
        else:
            current_word += ch
    # check the last word after the loop ends
    if count_characters(current_word) > count_characters(longest):
        longest = current_word
    return longest

text = input("Enter a sentence: ")
print("Longest word:", find_longest_word(text))
```

### 28. Count Occurrence of Each Vowel

```python
def count_each_vowel(text):
    a = e = i = o = u = 0
    for ch in text.lower():
        if ch == "a":
            a += 1
        elif ch == "e":
            e += 1
        elif ch == "i":
            i += 1
        elif ch == "o":
            o += 1
        elif ch == "u":
            u += 1
    print(f"a = {a}")
    print(f"e = {e}")
    print(f"i = {i}")
    print(f"o = {o}")
    print(f"u = {u}")

text = input("Enter text: ")
count_each_vowel(text)
```

## Section D – Loops, Patterns and Menus

### 29. Star Pattern

```python
def star_pattern(n):
    for i in range(1, n + 1):
        print("*" * i)

n = int(input("Enter number of rows: "))
star_pattern(n)
```

### 30. Number Pattern

```python
def number_pattern(n):
    for i in range(1, n + 1):
        row = ""
        for j in range(1, i + 1):
            row += str(j)
        print(row)

n = int(input("Enter number of rows: "))
number_pattern(n)
```
